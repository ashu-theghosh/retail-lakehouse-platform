```python
from pyspark.sql.functions import *
```

```python
crm_bronze=spark.read.format("delta").load("s3://retail-lakehouse-ashu/bronze/crm/")
```

```python
crm_bronze.groupBy(col("customer_id")).count().filter(col("count") > 1).display()
```

```python
crm_bronze.groupBy(col("customer_id").alias("customer_id")) \
    .agg(
        countDistinct("name.first_name").alias("first_name_count"),
        countDistinct("name.last_name").alias("last_name_count"),
        countDistinct("email").alias("email_count"),
        countDistinct("phone").alias("phone_count")
    ) \
    .filter((col("first_name_count") > 1) | (col("last_name_count") > 1) | (col("email_count") > 1) | (col("phone_count") > 1)) \
    .show()
```

```python
crm_bronze.groupBy("customer_id").agg(
    countDistinct("address.city").alias("city_count"),
    countDistinct("address.state").alias("state_count"),
    countDistinct("address.pincode").alias("pincode_count")
).filter(
    (col("city_count") > 1) |
    (col("state_count") > 1) |
    (col("pincode_count") > 1)
).show()
```

```python
crm_bronze.groupBy("customer_id").agg(
    countDistinct("loyalty.points").alias("points_count"),
    countDistinct("loyalty.tier").alias("tier_count")
).filter(
    (col("points_count") > 1) |
    (col("tier_count") > 1)
).show()
```

```python
crm_bronze.printSchema()
```

```python
crm_bronze.select(col("customer_id"),col("name.first_name"),col("name.last_name"),col("gender"),col("email"),col("phone"),col("address.city"),col("address.state"),col("address.pincode"),col("registration_date"),col("loyalty.points"),col("loyalty.tier"),col("registration_date"),col("ingestion_timestamp"),col("batch_id")).display()
```
